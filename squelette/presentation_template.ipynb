```python
import pandas as pd
import matplotlib.pyplot as plt

# Charge un dataset public (au choix). Ex Palmer Penguins :
URL = "https://raw.githubusercontent.com/mohan13krishna/Student-Academic-Performance-Dropout-Risk/main/student_dropout_risk_2026.csv"
df = pd.read_csv(URL)
df.head()
df.describe(include="all")
```

```python
df.groupby("nationality")["age"].mean().plot.bar(
    title="Age Moyen par nationalité"
)
plt.ylabel("Age")
plt.show()
print(df.groupby("nationality")["age"].agg(["mean", "std", "count"]).sort_values("mean"))
```

## 5. Réponses aux 3 questions d'acculturation IA

### 5.1 Apprentissage supervisé vs non supervisé

Supervisé se fait avec des étiquettes, on l'entraîne en connaissant la réponse. En biologie on pourrait imaginer identifier des espèces connues ?

Non supervisé n'utilise pas d'étiquettes, c'est à lui de découvrir la mécanique/pattern. Pour rester sur la biologie, regrouper des patients selon leur profil génique pour découvrir des sous-types de maladies inconnues ? (toujours dans l'idée d'identifier une anomalie qui sort du lot)

### 5.2 À quoi sert le transfer learning ?

*Réentrainer un modèle existant déjà entraîné sur un gros corpus proche du besoin qu'on adapte en réentraînant les dernières couches avec des données plus précises. Toujours sur la biologie : on peut prendre un grand modèle qui est entraîné sur l'identification par images des fleurs, puis le réentraîner avec des photos d'orchidées pour avoir un modèle capable de les différencier et spécialisé dessus. Cela évite de partir de 0 en faisant un modèle de A à Z (coût, temps, compétences requises)*

### 5.3 Pourquoi un LLM n'est pas toujours la bonne réponse ?

Le LLM est adapté à la compréhension de texte libre et pour le raisonement, mais pas une bonne solution pour des cas plus mathématiques tel que prédictif, binaire et peu de données sur lesquelles il peut appuyer sa reflexion.
